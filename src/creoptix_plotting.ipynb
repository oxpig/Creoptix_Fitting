## _Action_: Define file/folder locations in this cell

The only information that *must* be updated with each new Creoptix Project plot run is the path to the compressed Creoptix project `zipped_project_file` (the *.cxw file).  The path to this will define the parent directory where the following will reside.

The name of the project `project_name` is an arbitrary label for the user, but will have the date appended to help with organising.

Both `data_folder_name` and `results_folder_name` are also arbitrary:
 - `data_folder_name` specifies a *temporary* folder (with the `project_name` and date appended) where the unzipped project data will reside while the plotting runs.  If the plotting is successful, this folder and its contents will be deleted at the end of the run.
 - `results_folder_name` specifies a persistent folder (with the `project_name` and date appended) where the plotting results and summary table will be created.

 e.g. if the `zipped_project_file` is at `/path/to/project/SomeCreoptixRun.cxw`, then both the `data_folder_name` and `results_folder_name` folders will be generated inside `/path/to/project/`.

```python
 # change the below parts in quotes
project_name = "TestProject"    # appears in the plot titles
# Below is the path to the ZIPPED (i.e. *.cxw) Creoptix project file you wish to plot
zipped_project_file = "C:/Users/stat0524-admin/Creoptix_Fitting/data/20250826_DENV-2 NS2B3 Binding Assay.cxw"
data_folder_name = "Unzipped_Data"  # a temporary folder for data
results_folder_name = "Plotting_Results"    # folder where results go
```
